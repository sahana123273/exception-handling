2.What is exception handling in Python?
->A mechanism to handle runtime errors using try, except, else, and finally.

3.Purpose of finally block in exception handling?
->Runs always (with or without error). Used for cleanup (closing files, releasing resources).

4.What is logging in Python?
->Recording program information, warnings, and errors into console or files.

5.Significance of del method in Python?

->import module → use module.function()

from module import function → directly use function()

7.How can you handle multiple exceptions in Python?
->Using:

except (TypeError, ValueError):

OR multiple except blocks.

8.What is the Purpose of with statement in file handling
Automatically opens and closes files safely, even during errors.

9.Difference between multithreading and multiprocessing
->Multithreading: Multiple threads share the same memory.

Multiprocessing: Each process has its own memory.

10.What are the Advantages of using logging
->Helps in debugging

Records errors

Provides runtime monitoring

Maintains logs for future reference

11.What is memory management in Python?
->Automatic allocation & deallocation of memory using Garbage Collection + Reference Counting.

12.What are the basic Steps involved in exception handling
->Code at risk → try

Error handler → except

Run when no error → else

Cleanup → finally

13.Why is memory management important?
->Prevents memory leaks and improves performance.

14.What is the Role of try and except
->try: Code that may cause error

except: Handles the error safely

15.How Python’s garbage collector works?

Reference counting

Cyclic garbage collector

16.What is the Purpose of else block in exception handling ?
->Runs only if no exception occurs.

17.What are the Common logging levels in python ?
->DEBUG, INFO, WARNING, ERROR, CRITICAL

multiprocessing → Cross-platform process creation module

```python
#5. How would you catch and handle division by zero error in Python?

try:
    x = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")
```

```python
#6. Write a Python program that logs an error message to a log file when a division by zero exception occurs.

import logging

logging.basicConfig(filename="error.log", level=logging.ERROR)

try:
    a = int(input("Enter a number: "))
    b = int(input("Enter another number: "))
    result = a / b
    print("Result:", result)

except ZeroDivisionError:
    logging.error("Division by zero occurred.")
    print("Error: Cannot divide by zero! (Logged in error.log)")
```

```python
#11. Write a Python program that uses a try-except block to handle an error when attempting to access a dictionary key that doesn’t exist.

data = {"name": "Shreya"}

try:
    print(data["age"])
except KeyError:
    print("Key does not exist!")
```

```python
#13. How would you check if a file exists before attempting to read it in Python?

import os

if os.path.exists("demo.txt"):
    print("File exists")
else:
    print("File does not exist")
```

```python
#14. Write a program that uses the logging module to log both informational and error messages.

import logging

logging.basicConfig(filename="events.log", level=logging.INFO)

logging.info("Program started")

try:
    x = 10 / 0
except ZeroDivisionError as e:
    logging.error("Error occurred: %s", e)
    print("Logged the error in events.log")
```

```python
#15. Write a Python program that prints the content of a file and handles the case when the file is empty.

try:
    with open("demo.txt") as f:
        content = f.read()
        if content == "":
            print("File is empty")
        else:
            print(content)
except FileNotFoundError:
    print("File not found")
```

```python
#17. Write a Python program to create and write a list of numbers to a file, one number per line.

numbers = [10, 20, 30, 40, 50]

with open("numbers.txt", "w") as f:
    for n in numbers:
        f.write(str(n) + "\n")
```

```python
#18. How would you implement a basic logging setup that logs to a file with rotation after 1MB?

import logging
from logging.handlers import RotatingFileHandler

handler = RotatingFileHandler("mylog.log", maxBytes=1*1024*1024, backupCount=2)
logging.basicConfig(handlers=[handler], level=logging.INFO)

logging.info("Log rotation started")
```

```python
#22. How can you check if a file is empty before attempting to read its contents?

import os

size = os.path.getsize("demo.txt")

if size == 0:
    print("File is empty")
else:
    print("File is not empty")
```
